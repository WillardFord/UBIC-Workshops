# Guide to using AlphaFold



### 1. Navigate to AlphaFold Server

AlphaFold server is an online platform by Google DeepMind that hosts the AI system that predicts protein structure based on input sequences. AlphaFold3 can be used to show the structures of the protein and how they interact with other molecules.


To access the AlphaFold server, you will need a Google Account to sign in and save the jobs you run. You are given a number of jobs you can run per day that can be modified later.

AlphaFold Server:
https://alphafoldserver.com/

AlphaFold Database(find models of proteins AlphaFold predicts):
https://alphafold.ebi.ac.uk/

### 2. Get your input sequence

You will need a sequence of the protein you want AlphaFold to predict. This can be a novel sequence, or one that comes from an existing protein or residue.

You can find the sequences for protein models on the UniProt database, or the Protein Data Bank. Search up the name of the protein you or interested in, or identify the protein's PDB/UniProt ID.


UniProt: https://www.uniprot.org/

PDB: https://www.rcsb.org/

Example: ubiquitin aka 1UBQ

https://www.rcsb.org/structure/1ubq

### 3. Extra Options

The AlphaFold server does not allow for more parameters to be changed, but you can add more subunits with extra protein sequences. You can also add DNA sequences, RNA sequences, ligands, and ions for the protein to interact with in the prediction. Post translational modifications can also be applied to the sequence by editing it afterwards.

## 4. Run the Job

After the job looks ready to give, click "Continue and preview job". If  the job looks good, click "Confirm and submit job". This gives the job to AlphaFold to process, where the status should show as in progress.

## 5. Analyze the Results



AlphaFold viewer or ChimeraX can be used to see the results of the model in a 3D view, while comparing it to its reference model.

## Try Running Your own Protein Jobs

You can try running AlphaFold on one of the proteins below, or try picking your own protein!

NOTE: The time it takes for a job to complete on AlphaFold can take anywhere from a couple minutes to an entire day, depending on the length of the sequence, the optimization of the program, familiarity of the protein, and the resources you are using with your computer.

Smaller residues will take a shorter amount of time, as well as using GPU to make processes faster.
